Grayscale to binary.

Sample: `sample_butterfly.jpg`. The wings are dark and the petals are light, so the threshold is easy to see. Case 1 uses the mean intensity. Case 2 uses threshold 120.


```python
%matplotlib inline
import cv2
import numpy as np
import matplotlib.pyplot as plt


# grayscale -> binary
# case 1: threshold is the mean intensity
# case 2: threshold is typed in

def to_binary(img, thresh):
    out = np.zeros_like(img)
    out[img >= thresh] = 255
    return out
```

```python
image_path = "sample_butterfly.jpg"
user_threshold = 120

img = cv2.imread(image_path, 0)
print("sample image:", image_path, img.shape)

mean_t = float(img.mean())
bin_mean = to_binary(img, mean_t)
bin_user = to_binary(img, user_threshold)
print("Case 1, mean threshold =", round(mean_t, 2))
print("Case 2 threshold =", user_threshold)

plt.figure(figsize=(10, 4))
plt.subplot(1, 3, 1)
plt.imshow(img, cmap="gray", vmin=0, vmax=255)
plt.title("Original")
plt.axis("off")
plt.subplot(1, 3, 2)
plt.imshow(bin_mean, cmap="gray", vmin=0, vmax=255)
plt.title("Binary (mean)")
plt.axis("off")
plt.subplot(1, 3, 3)
plt.imshow(bin_user, cmap="gray", vmin=0, vmax=255)
plt.title("Binary (threshold 120)")
plt.axis("off")
plt.tight_layout()
plt.show()
```
